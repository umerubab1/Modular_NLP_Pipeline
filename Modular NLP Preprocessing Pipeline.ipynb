```python
import re
import string
import pandas as pd
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer, PorterStemmer
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
class TextPreprocessor:

    def __init__(self):
        self.stop_words = set(stopwords.words('english'))
        self.lemmatizer = WordNetLemmatizer()
        self.stemmer = PorterStemmer()

    #Cleaning
    def clean_text(self, text):
        if not isinstance(text, str):
            return ""

        text = text.lower()
        text = re.sub(r'\d+', '', text)  
        text = re.sub(f"[{string.punctuation}]", "", text)
        text = re.sub(r'\s+', ' ', text).strip()

        return text

    #Custom Tokenization 
    def tokenize(self, text):
        tokens = re.findall(r'\b\w+\b', text)
        return tokens

    #Stopword Removal
    def remove_stopwords(self, tokens):
        return [word for word in tokens if word not in self.stop_words]

    #Lemmatization
    def lemmatize(self, tokens):
        return [self.lemmatizer.lemmatize(word) for word in tokens]

    #Stemming
    def stem(self, tokens):
        return [self.stemmer.stem(word) for word in tokens]

    #Full Pipeline
    def preprocess(self, text, use_stemming=False, show_steps=False):

        cleaned = self.clean_text(text)
        tokens = self.tokenize(cleaned)
        no_stopwords = self.remove_stopwords(tokens)

        if use_stemming:
            final_tokens = self.stem(no_stopwords)
        else:
            final_tokens = self.lemmatize(no_stopwords)

        if show_steps:
            print("Original:", text)
            print("Cleaned:", cleaned)
            print("Tokens:", tokens)
            print("After Stopwords Removal:", no_stopwords)
            print("Final Output:", final_tokens)

        return final_tokens
```

```python
processor = TextPreprocessor()
text = "I am loving the new AI tools! They are running amazingly well."
output = processor.preprocess(text, show_steps=True)
print("\nProcessed Output:", output)
```

```python
df = pd.read_csv("/content/drive/MyDrive/sentiment_data.csv")  # column: Text
processor = TextPreprocessor()
df['Comment'] = df['Comment'].apply(lambda x: processor.preprocess(x))
df.to_csv("processed_dataset.csv", index=False)
print("Dataset processed and saved!")
```

```python
def menu():
    processor = TextPreprocessor()

    while True:
        print("\n--- Text Preprocessing System ---")
        print("1. Single Input")
        print("2. Batch Processing (CSV)")
        print("3. Exit")

        choice = input("Enter choice: ")

        if choice == '1':
            text = input("Enter text: ")
            result = processor.preprocess(text, show_steps=True)
            print("Final Output:", result)

        elif choice == '2':
            file = input("Enter CSV file path: ")

            df = pd.read_csv(file)
            column = input("Enter column name: ")

            df['processed'] = df[column].apply(lambda x: processor.preprocess(x))

            df.to_csv("processed_output.csv", index=False)
            print("Saved as processed_output.csv")

        elif choice == '3':
            break

        else:
            print("Invalid choice!")

menu()
```
